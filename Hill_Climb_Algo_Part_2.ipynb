```python

from itertools import product

def implies(a, b):
    return not a or b

def check_kb(P, Q, R):
    s1 = implies(Q, P)       # Q -> P
    s2 = implies(P, not Q)   # P -> NOT Q
    s3 = Q or R              # Q OR R

    return s1, s2, s3

print("P Q R | Q->P P->~Q Q|R | KB")
print("--------------------------------")

models = []

for P, Q, R in product([False, True], repeat=3):
    s1, s2, s3 = check_kb(P, Q, R)
    kb = s1 and s2 and s3

    print(int(P), int(Q), int(R), "| ",
          int(s1), "   ", int(s2), "   ",
          int(s3), " | ", kb)

    if kb:
        models.append((P, Q, R))

print("\nModels of KB:", models)

entails_R = all(R for P, Q, R in models)
entails_R_implies_P = all(implies(R, P) for P, Q, R in models)
entails_Q_implies_R = all(implies(Q, R) for P, Q, R in models)

print("KB entails R:", entails_R)
print("KB entails R -> P:", entails_R_implies_P)
print("KB entails Q -> R:", entails_Q_implies_R)
```
